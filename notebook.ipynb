```python
import pandas as pd
from datetime import date
```

```python
df = pd.read_csv("reactivos_comercial.csv")
```

```python
df.head(1).T
```

```python
def add_units(row):
    if row["estado"] == "Sólido":
        return f"{row.cantidad} g"
    elif row["estado"] == "Líquido":
        return f"{row["cantidad"]} mL"
    else:
        return row["cantidad"]
```

```python
df.loc[:, ["cantidad", "estado"]].apply(add_units, axis = 0)
```

Conversión a datetime de la columna `fecha_caducidad`

```python
df.fecha_caducidad = pd.to_datetime(df.fecha_caducidad).dt.date
```

Creación de nuevas columnas

```python
df["valor"] = round(df.cantidad / 100 * df.precio_100, 2)
df["bajo_stock"] = df.cantidad < df.stock_minimo
df["caducado"] = df.fecha_caducidad > date.today()
```

Renombrado de columnas de archivos

```python
df = df.rename(columns = {"ficha_pdf" : "ficha_url",
                          "pictograma" : "picto_url"})
```

Generación de la ruta de los archivos

```python
df.ficha_url = "static/" + df.ficha_url
df.picto_url = "static/" + df.picto_url
```
